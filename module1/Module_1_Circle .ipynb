```python
import math

#Class Definition
class Circle:

    #define the Constructor
    def __init__(self, radius):
        self.radius = radius

    #define the method area and returns the area of the circle.
    def area(self):
        return math.pi * self.radius ** 2

    #define the method circumference and returns the circumference of the circle.
    def circumference(self):
        return 2 * math.pi * self.radius

    #define the method diameter and returns the diameter of the circle.
    def diameter(self):
        return 2 * self.radius

    #define the method is_unit_circle and check if is unit circle, then returns True if the circle is a unit circle (radius = 1), otherwise False.
    def is_unit_circle(self):
        return self.radius == 1

    #Displays the circle information
    def display(self):
        print("area: ", self.area())
        print("circumference: ", self.circumference())
        print("diameter: ", self.diameter())
        print("is unit circle?:", self.is_unit_circle())

    #End of class
    #Create three objects of the class circle and put it to the list
circles = []

radius1 = float(input("Enter the radius of circle 1: "))
circles.append(Circle(radius1))

radius2 = float(input("Enter the radius of circle 2: "))
circles.append(Circle(radius2))

radius3 = float(input("Enter the radius of circle 3: "))
circles.append(Circle(radius3))

#Call the display method
for circle in circles:
    circle.display()
    print()
```
